````python
import streamlit as st
import pandas as pd

st.title("Dashboard de Telemetría Deportiva - Bicicleta")
st.write("Análisis de datos experimentales del recorrido en bicicleta")

try:
    df = pd.read_csv('datos_simulados.csv', sep=';')

    # Calcular Posición de forma aproximada usando la integral de la velocidad respecto al tiempo (cumprod/cumsum)
    # Asumiendo que el paso de tiempo medio es constante (dt = Tiempo_i - Tiempo_{i-1})
    dt = df['Tiempo'].diff().fillna(0)
    df['Posicion'] = (df['Velocidad'] * dt).cumsum()

    # Calcular Fuerza Neta (Masa = 85 kg)
    df['Fuerza_Neta'] = 85 * df['Aceleracion']

    st.subheader("📊 Datos Crudos del Experimento")
    st.dataframe(df.head())

    st.subheader("📍 Posición vs. Tiempo")
    st.line_chart(df.set_index('Tiempo')['Posicion'])

    st.subheader("📈 Velocidad vs. Tiempo")
    st.line_chart(df.set_index('Tiempo')['Velocidad'])

    st.subheader("🚀 Aceleración vs. Tiempo")
    st.line_chart(df.set_index('Tiempo')['Aceleracion'])

    st.subheader("⚡ Fuerza Neta vs. Tiempo")
    st.line_chart(df.set_index('Tiempo')['Fuerza_Neta'])

    st.success("✅ Datos procesados y las tres gráficas se visualizaron correctamente.")

except FileNotFoundError:
    st.error("⚠️ No se encontró el archivo 'datos_simulados.csv'. Asegúrate de subirlo a la carpeta de Colab.")
````

````python
# 1. Creamos el archivo app.py con diseño profesional, KPIs de resumen y gráficos en una línea
with open('app.py', 'w', encoding='utf-8') as f:
    f.write('''import streamlit as st
import pandas as pd
import os

# Configuración de página ancha para visualización profesional
st.set_page_config(
    page_title="Dashboard de Telemetría Deportiva",
    page_icon="🚴",
    layout="wide"
)

# Estilo CSS personalizado para mejorar las fuentes y bordes
st.markdown("""
    <style>
    .main-title { font-size: 2.2rem; font-weight: bold; color: #1E3A8A; margin-bottom: 0.5rem; }
    .sub-title { font-size: 1.1rem; color: #4B5563; margin-bottom: 2rem; }
    </style>
""", unsafe_allow_html=True)

st.markdown('<div class="main-title">🚴 Dashboard de Telemetría Deportiva — Ciclismo</div>', unsafe_allow_html=True)
st.markdown('<div class="sub-title">Análisis de variables cinemáticas y dinámicas obtenidas en tiempo real durante el experimento de ruta.</div>', unsafe_allow_html=True)

ruta_archivo = '/content/datos_simulados.csv'

if not os.path.exists(ruta_archivo):
    st.error(f"⚠️ No se encontró el archivo en la ruta esperada: {ruta_archivo}. Por favor, asegúrate de haberlo subido a Colab.")
else:
    try:
        df = pd.read_csv(ruta_archivo, sep=';')

        # Cálculos de física e ingeniería
        dt = df['Tiempo'].diff().fillna(0)
        df['Posicion'] = (df['Velocidad'] * dt).cumsum()
        df["Fuerza_Neta"] = 85.0 * df["Aceleracion"]  # Asumiendo masa del sistema ciclista + bicicleta = 85 kg

        # Sección 1: Tarjetas de Métricas Clave (KPIs)
        st.subheader("📊 Resumen del Recorrido (Métricas de Rendimiento)")
        kpi1, kpi2, kpi3, kpi4 = st.columns(4)

        distancia_total = df['Posicion'].iloc[-1]
        vel_maxima = df['Velocidad'].max()
        aceleracion_max = df['Aceleracion'].max()
        fuerza_max = df['Fuerza_Neta'].max()

        kpi1.metric(label="📍 Distancia Total Recorrida", value=f"{distancia_total:.2f} m", delta="Trayecto")
        kpi2.metric(label="📈 Velocidad Máxima", value=f"{vel_maxima:.2f} m/s", delta=f"{vel_maxima * 3.6:.1f} km/h")
        kpi3.metric(label="🚀 Aceleración Máxima", value=f"{aceleracion_max:.2f} m/s²")
        kpi4.metric(label="⚡ Fuerza Máxima ejercida", value=f"{fuerza_max:.1f} N", delta="Masa: 85 kg")

        st.divider()

        # Sección 2: Las 4 Gráficas en una sola línea de manera profesional
        st.subheader("📈 Gráficas Simultáneas de Física del Movimiento")

        col1, col2, col3, col4 = st.columns(4)

        with col1:
            st.markdown("#### 📍 Posición ($x$ vs $t$)")
            st.caption("Cambio de posición espacial acumulada en metros ($m$).")
            st.line_chart(df.set_index("Tiempo")["Posicion"])

        with col2:
            st.markdown("#### 📈 Velocidad ($v$ vs $t$)")
            st.caption("Velocidad instantánea medida en metros por segundo ($m/s$).")
            st.line_chart(df.set_index("Tiempo")["Velocidad"])

        with col3:
            st.markdown("#### 🚀 Aceleración ($a$ vs $t$)")
            st.caption("Ritmo de cambio de la velocidad medido en $m/s^2$.")
            st.line_chart(df.set_index("Tiempo")["Aceleracion"])

        with col4:
            st.markdown("#### ⚡ Fuerza Neta ($F$ vs $t$)")
            st.caption("Fuerza resultante aplicada calculada mediante $F = m \\cdot a$ ($N$).")
            st.line_chart(df.set_index("Tiempo")["Fuerza_Neta"])

        st.divider()

        # Sección 3: Datos de respaldo interactivos
        with st.expander("🗂️ Ver la tabla completa de datos crudos procesados"):
            st.dataframe(df, use_container_width=True)

    except Exception as e:
        st.error(f"⚠️ Ocurrió un error al procesar el archivo: {e}")
''')
print("Archivo 'app.py' actualizado con éxito con diseño de 4 columnas en una sola línea. Vuelve a ejecutar la celda 3 (iniciar Streamlit y Ngrok) para ver la nueva versión.")
````
